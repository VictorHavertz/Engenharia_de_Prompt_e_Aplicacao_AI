criar um programa que organize arquivos bagunçados em pastas automaticamente, baseandi-se apenas na extenção do arquivo(ex:pdf,jpg). Use a biblioteca os ou shuitil

```python
# Importa as bibliotecas necessárias para manipulação de arquivos e diretórios.
import os
import shutil

# Define a função principal para organizar os arquivos.
def organizar_arquivos(diretorio_origem):
    """
    Organiza arquivos no diretório de origem fornecido em subpastas
    baseadas em suas extensões de arquivo.
    """
    # Verifica se o diretório de origem existe. Se não existir, imprime um erro e sai.
    if not os.path.isdir(diretorio_origem):
        print(f"Erro: O diretório '{diretorio_origem}' não existe.")
        return

    print(f"Iniciando a organização de arquivos em: {diretorio_origem}")

    # Itera sobre cada item (arquivo ou diretório) no diretório de origem.
    for item in os.listdir(diretorio_origem):
        # Constrói o caminho completo para o item.
        caminho_item = os.path.join(diretorio_origem, item)

        # Verifica se o item é um arquivo (e não um diretório).
        if os.path.isfile(caminho_item):
            # Separa o nome do arquivo da sua extensão.
            nome_arquivo, extensao_arquivo = os.path.splitext(item)

            # Remove o ponto inicial da extensão e converte para minúsculas para padronização.
            # Se a extensão for '.pdf', a pasta será 'pdf'.
            nome_pasta = extensao_arquivo[1:].lower()

            # Se o arquivo não tiver extensão, ele é movido para uma pasta chamada 'outros'.
            if not nome_pasta:
                nome_pasta = 'outros'

            # Constrói o caminho completo para a pasta de destino (dentro do diretório de origem).
            pasta_destino = os.path.join(diretorio_origem, nome_pasta)

            # Cria a pasta de destino se ela ainda não existir (exist_ok=True evita erro se já existir).
            os.makedirs(pasta_destino, exist_ok=True)

            # Tenta mover o arquivo para a pasta de destino.
            try:
                shutil.move(caminho_item, pasta_destino)
                print(f"Movido '{item}' para '{pasta_destino}'")
            # Captura exceções se houver um erro ao mover o arquivo (por exemplo, permissão negada).
            except shutil.Error as e:
                print(f"Erro ao mover '{item}': {e}")
        # Se o item for um diretório, ele é ignorado.
        elif os.path.isdir(caminho_item):
            print(f"Pulando diretório: '{item}'")

    print("Organização de arquivos concluída!")

# --- Exemplo de Uso ---

# Define o diretório a ser organizado.
# Para demonstração, usaremos uma subpasta dentro de '/content/'.
# Você pode alterar 'diretorio_base' para o caminho desejado em seu sistema.
diretorio_base = '/content/arquivos_organizados_demo'

# Cria o diretório base se ele não existir.
os.makedirs(diretorio_base, exist_ok=True)

# Cria alguns arquivos fictícios para teste e demonstração.
print("\nCriando arquivos fictícios para demonstração...")
arquivos_exemplo = {
    'documento.pdf': 'Este é um documento PDF de exemplo.',
    'imagem.jpg': 'Esta é uma imagem JPG de exemplo.',
    'relatorio.docx': 'Este é um documento do Word de exemplo.',
    'dados.csv': 'coluna1,coluna2\n1,2\n3,4',
    'notas.txt': 'Algumas notas em texto simples.',
    'outra_imagem.png': 'Outro arquivo de imagem PNG.',
    'apresentacao.pptx': 'Uma apresentação PowerPoint.',
    'arquivo.zip': 'Um arquivo compactado.',
    'arquivo_sem_extensao': 'Este arquivo não possui extensão.'
}

# Escreve o conteúdo nos arquivos fictícios dentro do diretório base.
for nome_arquivo, conteudo in arquivos_exemplo.items():
    caminho_arquivo = os.path.join(diretorio_base, nome_arquivo)
    with open(caminho_arquivo, 'w') as f:
        f.write(conteudo)
    print(f"Criado '{nome_arquivo}'")

# Agora, chama a função para organizar os arquivos no diretório especificado.
print("\nIniciando o processo de organização...")
organizar_arquivos(diretorio_base)

print(f"\nDemonstração finalizada. Verifique o diretório '{diretorio_base}' para ver os arquivos organizados.")
```

desenvolver um script que consulte uma API publica (viaCep) e retorne as informações de endereço formatados de maneira legivel. inclua tratamento de execuções.

```python
# Importa a biblioteca 'requests' para fazer requisições HTTP (para a API).
import requests

def consultar_cep(cep):
    """
    Consulta a API ViaCEP para obter informações de endereço de um CEP.
    Inclui tratamento de exceções e formatação legível.
    """
    # Remove quaisquer caracteres não numéricos do CEP para garantir o formato correto.
    cep = str(cep).replace("-", "").replace(".", "").replace(" ", "")

    # Verifica se o CEP tem 8 dígitos, que é o formato padrão brasileiro.
    if len(cep) != 8 or not cep.isdigit():
        return {"erro": "CEP inválido. Deve conter 8 dígitos numéricos."}

    # Constrói a URL da API ViaCEP.
    url = f"https://viacep.com.br/ws/{cep}/json/"

    try:
        # Faz a requisição GET para a API.
        response = requests.get(url)
        response.raise_for_status()  # Levanta uma exceção para códigos de status HTTP de erro (4xx ou 5xx).

        # Converte a resposta JSON em um dicionário Python.
        dados_endereco = response.json()

        # Verifica se a API retornou um erro (ex: CEP não encontrado).
        if 'erro' in dados_endereco and dados_endereco['erro'] == True:
            return {"erro": f"CEP '{cep}' não encontrado ou inexistente."}

        # Formata as informações de endereço de maneira legível.
        endereco_formatado = {
            "CEP": dados_endereco.get('cep'),
            "Logradouro": dados_endereco.get('logradouro'),
            "Complemento": dados_endereco.get('complemento'),
            "Bairro": dados_endereco.get('bairro'),
            "Cidade": dados_endereco.get('localidade'),
            "Estado": dados_endereco.get('uf'),
            "IBGE": dados_endereco.get('ibge'),
            "GIA": dados_endereco.get('gia'),
            "DDD": dados_endereco.get('ddd'),
            "SIAFI": dados_endereco.get('siafi')
        }

        return endereco_formatado

    except requests.exceptions.HTTPError as http_err:
        return {"erro": f"Erro HTTP: {http_err} (Status: {response.status_code})"}
    except requests.exceptions.ConnectionError as conn_err:
        return {"erro": f"Erro de Conexão: {conn_err}. Verifique sua conexão com a internet."}
    except requests.exceptions.Timeout as timeout_err:
        return {"erro": f"Tempo limite excedido: {timeout_err}. A API demorou muito para responder."}
    except requests.exceptions.RequestException as req_err:
        return {"erro": f"Erro na requisição: {req_err}"}
    except ValueError:
        return {"erro": "Erro ao decodificar a resposta JSON. A API pode ter retornado dados inválidos."}
    except Exception as e:
        return {"erro": f"Ocorreu um erro inesperado: {e}"}

# --- Exemplos de Uso ---
```

```python
# Permite ao usuário digitar um CEP para consulta interativa.
print("\n--- Consulta de CEP Interativa ---")
cep_digitado = input("Por favor, digite um CEP (apenas números ou com hífen): ")

resultado_interativo = consultar_cep(cep_digitado)

if "erro" in resultado_interativo:
    print(f"Erro na consulta: {resultado_interativo['erro']}")
else:
    print("\n--- Informações do Endereço ---")
    for chave, valor in resultado_interativo.items():
        # Garante que 'None' seja exibido como vazio ou 'Não disponível' para melhor legibilidade
        valor_formatado = valor if valor is not None and valor != '' else 'Não disponível'
        print(f"{chave}: {valor_formatado}")

print("\n--- Fim da Consulta Interativa ---")
```
